```python
import sqlite3
import pandas as pd

# 1. Crear conexión en memoria y cursor
conexion = sqlite3.connect(":memory:")
cursor = conexion.cursor()

# 2. Crear tabla de ventas
cursor.execute("""
CREATE TABLE ventas_tecnologia (
    id_venta INTEGER PRIMARY KEY,
    producto TEXT,
    categoria TEXT,
    precio_unitario REAL,
    cantidad INTEGER,
    fecha TEXT,
    pais TEXT
);
""")

# 3. Insertar datos de prueba
datos_ejemplo = [
    (1, 'Laptop Pro 15', 'Computadoras', 1200.0, 10, '2024-01-15', 'Colombia'),
    (2, 'Mouse Ergonómico', 'Accesorios', 35.0, 50, '2024-01-16', 'Argentina'),
    (3, 'Monitor 4K', 'Pantallas', 650.0, 8, '2024-01-18', 'Colombia'),
    (4, 'Teclado Mecánico', None, 95.0, 20, '2024-01-20', 'Chile'),
    (5, 'Smartphone Ultra', 'Telefonía', 900.0, 15, '2024-01-22', 'Colombia'),
    (6, 'Cable HDMI', 'Accesorios', 15.0, 100, '2024-01-25', 'Colombia'),
    (7, 'Tablet 10 pulgadas', None, 400.0, 5, '2024-01-28', 'Perú')
]

cursor.executemany("INSERT INTO ventas_tecnologia VALUES (?, ?, ?, ?, ?, ?, ?)", datos_ejemplo)
conexion.commit()

# Función auxiliar para consultar y mostrar como DataFrame
def ejecutar_consulta(query):
    return pd.read_sql_query(query, conexion)
```
